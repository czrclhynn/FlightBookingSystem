## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains a large number of pages and shows that a meaningful share of them have declining performance. It also contains pages that have not been updated for a long time but still receive substantial impressions. These patterns suggest that content-review prioritization is a useful problem to investigate because teams may benefit from identifying which pages deserve attention first instead of reviewing every page equally.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

total_pages = len(df)

declining_rate = (
    df["trend_direction"]
    .str.lower()
    .eq("down")
    .mean()
)

stale_visible = (
    (df["days_since_last_update"] >= 180) &
    (df["impressions_90d"] >= 500)
)

stale_visible_count = stale_visible.sum()
stale_visible_rate = stale_visible.mean()

print(f"Total pages: {total_pages:,}")
print(f"Pages with a downward trend: {declining_rate:.1%}")
print(
    f"Pages at least 180 days since update and with at least 500 impressions: "
    f"{stale_visible_count:,} ({stale_visible_rate:.1%})"
)
```
